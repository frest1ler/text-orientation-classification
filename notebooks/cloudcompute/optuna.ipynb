# CloudCompute: Optuna
Изолированный поиск гиперпараметров MobileNet с возобновляемой SQLite study.

```python
REPO_URL = "https://github.com/frest1ler/text-orientation-classification.git"
BRANCH = "main"
SMOKE_RUN = True
TRIALS = 10
TRAIN_BASE_SAMPLES = 8000
VALIDATION_BASE_SAMPLES = 5000
NUM_WORKERS = 2
RUN_TESTS = False
REPO_DIR = "/root/text-orientation-classification"
STATE_DIR = "/root/text-orientation-state"
```

```python
import os, subprocess, sys
from pathlib import Path
repo = Path(REPO_DIR)
if not (repo / ".git").is_dir(): subprocess.run(["git", "clone", "--branch", BRANCH, REPO_URL, str(repo)], check=True)
os.chdir(repo)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
if RUN_TESTS: subprocess.run([sys.executable, "-m", "pytest", "-q"], check=True)
Path(STATE_DIR).mkdir(parents=True, exist_ok=True)
```

```python
command = [sys.executable, "-m", "scripts.optuna_search", "--project-dir", STATE_DIR, "--trials", str(TRIALS), "--train-base-samples", str(TRAIN_BASE_SAMPLES), "--validation-base-samples", str(VALIDATION_BASE_SAMPLES), "--num-workers", str(NUM_WORKERS)]
if SMOKE_RUN: command.append("--smoke")
subprocess.run(command, check=True)
result_dir = Path(STATE_DIR) / "tuning/optuna"
print("Persistent study and reports:", result_dir)
```
